```python
# write a program to calculate simple intrest and total amount by reading necessary inputs
p=float(input("enter principle amount"))
t=float(input("enter time period"))
r=float(input("enter rate of intrest"))
SI=(p*t*r)/100
ta=p+SI
print(f"the simple interest is {SI:.2f}Rs")
print(f"the total amount is {ta:.2f}Rs")
```
