```python
# Square hallow pattern
for i in range(1,7,1):
    for j in range(1,6,1):
        if i==1 or i==6 or j==1 or j==5:
            print("*", end=" ")
        else:
            print("  ",end="")
    print()
```

```python
# Top half of the butterfly (1 up to 4)
for i in range(1, 5, 1):
    # Left wing
    for j in range(1, i + 1, 1):
        print("*", end=" ")
    # Middle spaces
    for s in range(2 * (4 - i) - 1):
        print(" ", end=" ")
    # Right wing (skips 1 star on the center row)
    for j in range(1, i + 1, 1):
        if i == 4 and j == 4:
            continue
        print("*", end=" ")
    print()

# Bottom half of the butterfly (3 down to 1)
for i in range(3, 0, -1):
    # Left wing
    for k in range(1, i + 1, 1):
        print("*", end=" ")
    # Middle spaces
    for s in range(2 * (4 - i) - 1):
        print(" ", end=" ")
    # Right wing
    for k in range(1, i + 1, 1):
        print("*", end=" ")
    print()

```
