#### Python operators
It's a symbol used to perform a certain operation between two operands
Ex:
s=a+b

#### Types
1) arithmetic operators: +, -, *, /, %, //, **
2) comparison operators/ relational operators: ==, !=, >,<,>=,<= (gives true/false as output)
3) assignment operators: =, +=, -= ,*=, /=, %=, //=, **=
4) logical operators: and, or, not
5) membership operators: in, not in
