```python
def check_prime(num):
    if num < 2:
        return "Not a Prime Number"

    for i in range(2, num):
        if num % i == 0:
            return "Not a Prime Number"

    return "Prime Number"


num = int(input("Enter a number: "))
res = check_prime(num)
print(res)
```

```python
def sum_of_digits(num):
    s=0
    while num>0:
        rem=num%10
        s=s+rem
        num=num//10
    return s
num=int(input("enter a number:"))
res=sum_of_digits(num)
print(res)
```

```python
def check_palindrome(num):
    temp=num
    rev=0
    while temp>0:
        rem=temp%10
        rev=(rev*10)+rem
        temp=temp//10
    if rev==num:
        return "Palindrome"
    else:
        return "Not palindrome"
num=int(input("enter a number:"))
res=check_palindrome(num)
print(res)

```

```python
def check_armstrong(num):
    temp=num
    n1=str(num)
    n2=len(n1)
    s=0
    while temp>0:
        rem=temp%10
        s=s+rem**n2
        temp=temp//10
    if s==num:
        return "Armstrong Number"
    else:
        return "Not an Armstrong Number"
num=int(input("enter a num:"))
res=check_armstrong(num)
print(res)
```

```python
def count_even_odd(numbers):
    even_count=0
    odd_count=0
    for i in numbers:
        if i%2==0:
            even_count+=1
        else:
            odd_count+=1
    return even_count,odd_count
numbers=[10, 15, 20, 25, 30, 35, 40]
res=count_even_odd(numbers)
print(f"Even Numbers count={res[0]}\nOdd Numbers Count={res[1]}")

    
```

```python
def calculate_average(numbers):
    total=0
    count=0
    for i in numbers:
        total=total+i
        count+=1
    avg=total/count
    return avg
numbers=[80, 90, 75, 85, 70]
res=calculate_average(numbers)
print("Average=",res)
```

```python
def find_duplicates(numbers):
    l=[]
    duplicates=[]
    for i in numbers:
        if i not in l:
            l.append(i)
        elif i not in duplicates:
            duplicates.append(i)
    return duplicates
numbers=[10, 20, 30, 20, 40, 10, 50, 30]
res=find_duplicates(numbers)
print("duplicate elements=",res)

            
```

```python
def second_largest(numbers):
    largest=numbers[0]
    sec_largest=numbers[0]
    for i in numbers:
        if i>largest:
            sec_largest=largest
            largest=i
        elif i>sec_largest and i!=largest:
            sec_largest=i
            
    return sec_largest
numbers=[10, 50, 30, 80, 60, 80]
res=second_largest(numbers)
print(res)
```
