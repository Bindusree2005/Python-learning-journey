```python
# create an even num list and odd num list between 1 to 20
l1=[]
l2=[]
for i in range(1,20):
    if i%2==0:
        l1.append(i)
    else:
        l2.append(i)
print("even num",l1)
print("odd num",l2)
```

```python
# l2=[3,2,9,5,12,10] create square list of given list values 
l2=[3,2,9,5,12,10]
sl=[]
for i in l2:
    s=i**2
    sl.append(s)
print(sl)
```
